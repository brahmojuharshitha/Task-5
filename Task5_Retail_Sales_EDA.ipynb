# Task 5: Exploratory Data Analysis (EDA)\n## Retail Sales Dataset\nThis notebook explores a small practice retail-sales dataset. The dataset is synthetic and is used for learning EDA methods.

## 1. Import libraries and load data\nPandas is used for data handling, Matplotlib and Seaborn are used for charts.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset (keep retail_sales.csv in the same folder as this notebook)
df = pd.read_csv("retail_sales.csv")
df.head()
```

## 2. Understand the data\nThe shape, info and descriptive statistics help us understand the columns and values.

```python
# Basic information about the dataset
print("Shape:", df.shape)
df.info()
df.describe(include="all")
```

## 3. Data cleaning\nA few category and region entries are missing. I fill them with the most frequent value in each column.

```python
# Check missing values and remove duplicates
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

# Fill missing category and region values with the most common value
df["Category"] = df["Category"].fillna(df["Category"].mode()[0])
df["Region"] = df["Region"].fillna(df["Region"].mode()[0])
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Month"] = df["Order_Date"].dt.month_name()

print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 4. Univariate analysis: category sales

```python
# Sales by category
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
category_sales.plot(kind="bar", title="Total Sales by Category", ylabel="Sales", xlabel="Category")
plt.xticks(rotation=30)
plt.show()

# Observation: Compare the bars to identify the category with the highest sales.
```

## 5. Univariate analysis: region sales

```python
# Sales by region
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
region_sales.plot(kind="bar", title="Total Sales by Region", ylabel="Sales", xlabel="Region")
plt.show()

# Observation: The tallest bar represents the region with the highest total sales.
```

## 6. Distribution of sales

```python
# Distribution of sales values
plt.figure(figsize=(7, 4))
plt.hist(df["Sales"], bins=25, edgecolor="black")
plt.title("Distribution of Order Sales")
plt.xlabel("Sales per Order")
plt.ylabel("Number of Orders")
plt.show()

# Observation: The histogram shows how frequently different order values occur.
```

## 7. Bivariate analysis: units sold and sales

```python
# Relationship between units sold and sales
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Units_Sold", y="Sales", hue="Category")
plt.title("Units Sold vs Sales")
plt.show()

# Observation: Sales generally tend to increase when more units are sold.
```

## 8. Correlation analysis

```python
# Correlation heatmap for numerical columns
numeric_cols = ["Units_Sold", "Unit_Price", "Discount", "Sales"]
correlation = df[numeric_cols].corr()
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

# Observation: Values closer to 1 show a stronger positive relationship;
# values closer to -1 show a stronger negative relationship.
```

## 9. Final summary

```python
# Summary of important findings
print("Total sales:", round(df["Sales"].sum(), 2))
print("Average order sales:", round(df["Sales"].mean(), 2))
print("Highest sales category:", df.groupby("Category")["Sales"].sum().idxmax())
print("Highest sales region:", df.groupby("Region")["Sales"].sum().idxmax())
```

## Conclusion\nEDA helped identify sales patterns across categories and regions, understand the distribution of order values, and examine relationships between numerical columns. These findings can support further business analysis.
